# Step 3: Data Cleaning & Preprocessing

```python
# Creating a copy before performing a analysis
df_clean = df.copy()

df_clean.shape
```

3.1 Handling missing values using mean, median, or mode

```python
# check missing values in each column
missing_values = df_clean.isnull().sum()
print(missing_values)
```

```python
# Display only columns containing missing values
missing_values = missing_values[missing_values > 0]

print("Missing values in each column:")
print(missing_values.sort_values(ascending=False))
```

```python
# Calculate the percentage of missing values
missing_percentage = (df_clean.isnull().sum() / len(df_clean) * 100).round(2)

print("\nMissing value percentage:")
print(missing_percentage[missing_percentage > 0].sort_values(ascending=False))
```

```python
# Check missing values in Financial missing-value summary

financial_cols = [
    "Order_Value",
    "Discount_Applied",
    "Final_Amount"
]

print("Missing financial values:")
print(df_clean[financial_cols].isnull().sum())

print("\nRows with all three financial values missing:")
print(df_clean[financial_cols].isnull().all(axis=1).sum())
```

```python
 #Group remaining missing values

numeric_missing = df_clean.select_dtypes(include="number").isnull().sum()

categorical_missing = df_clean.select_dtypes(include=["object", "string", "bool"]
).isnull().sum()

print("Missing numeric values:")
print(numeric_missing[numeric_missing > 0])

print("\n Missing categorical values:")
print(categorical_missing[categorical_missing > 0])
```

First check the categorical values using mode

Check the cancellation reason --> 90969

```python

# Check cancellation reasons by order status

print("Order status distribution:")
print(df_clean["Order_Status"].value_counts())

print("\n Cancellation reason availability by order status:")
print(df_clean.groupby("Order_Status")["Cancellation_Reason"]
    .apply(lambda x: x.isnull().value_counts()))
```

Handle missing cancellation reasons

For delivered orders, we can label the missing reason as "Not Applicable". For cancelled orders with missing reasons, we should use "Unknown".

```python

# Handle missing cancellation reasons based on order status

delivered_mask = df_clean["Order_Status"] == "Delivered"
cancelled_mask = df_clean["Order_Status"] == "Cancelled"

# Missing reason for delivered orders: not applicable
df_clean.loc[ delivered_mask & df_clean["Cancellation_Reason"].isna(),
    "Cancellation_Reason"] = "Not Applicable"

# Missing reason for cancelled orders: unknown
df_clean.loc[cancelled_mask & df_clean["Cancellation_Reason"].isna(),
    "Cancellation_Reason"] = "Unknown"

# Verify
print("Remaining missing cancellation reasons:",
      df_clean["Cancellation_Reason"].isnull().sum())

print("\nCancellation reason distribution:")
print(df_clean["Cancellation_Reason"].value_counts())
```

Check missing values in financial, categorical, rating, and time-related columns. Let's handle the next group: Payment_Mode.

```python

# Check remaining categorical missing values

categorical_cols = [
    "Customer_Gender",
    "Payment_Mode",
    "City",
    "Area",
    "Cuisine_Type"
]

print(df_clean[categorical_cols].isnull().sum())
```

Filling unknown for missing categorical value

```python

# Fill missing categorical values with "Unknown"

categorical_cols = [
    "Customer_Gender",
    "Payment_Mode",
    "City",
    "Area",
    "Cuisine_Type"
]

for col in categorical_cols:
    df_clean[col] = df_clean[col].fillna("Unknown")

# Verify
print("Remaining missing categorical values:")
print(df_clean[categorical_cols].isnull().sum())
```

Check missing values in numerical columns

```python
missing_values = df_clean.isnull().sum()

print(missing_values[missing_values > 0].sort_values(ascending=False))
```

```python
print("Shape:", df_clean.shape)

print(df_clean[
    ["Customer_Age", "Distance_km", "Delivery_Time_Min", "Final_Amount"]
].isnull().sum())
```

Handling missing numerical columns using median

```python
# Fill missing values using the median
median_cols = [
    "Customer_Age",
    "Distance_km",
    "Delivery_Time_Min"
]

for col in median_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# Verify
print(df_clean[median_cols].isnull().sum())
```

Verifying the remining missing values

```python
missing_values = df_clean.isnull().sum()
print(missing_values[missing_values > 0].sort_values(ascending=False))
```

Checking the missing discounts

Discount_Applied  -->  16715

```python
# Check discount distribution
print(df_clean["Discount_Applied"].value_counts(dropna=False).sort_index())

# Check missing discounts by order status
print(pd.crosstab(
    df_clean["Order_Status"],
    df_clean["Discount_Applied"].isna(),
    margins=True
))
```

Checking Discount_applied with Order_value

```python
# Check whether missing discounts also have missing order values
print(pd.crosstab(
    df_clean["Discount_Applied"].isna(),
    df_clean["Order_Value"].isna(),
    margins=True
))
```

3.2 Removing or Capping Outliers (delivery time, order value)

Detect potential outliers

```python
outlier_cols = ["Delivery_Time_Min", "Order_Value"]

for col in outlier_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[col] < lower) |
        (df_clean[col] > upper)
    ]

    print(f"\nColumn: {col}")
    print("Lower bound:", lower)
    print("Upper bound:", upper)
    print("Number of outliers:", len(outliers))
```

```python
# Check the actual minimum and maximum

for col in ["Delivery_Time_Min", "Order_Value"]:
    print(f"\n{col}")
    print("Minimum:", df_clean[col].min())
    print("Maximum:", df_clean[col].max())
    print("Negative values:", (df_clean[col] < 0).sum())
```

```python
# Correct Invalid Values So check the values of Profit_Margin

print("Minimum Profit_Margin:", df_clean["Profit_Margin"].min())
print("Negative profit margins:", (df_clean["Profit_Margin"] < 0).sum())
```

Check negative profit margins

```python
# Check profit margin distribution
print(df_clean["Profit_Margin"].describe())

# Count negative, zero, and positive margins
print("\nMargin categories:")
print(pd.Series({
    "Negative": (df_clean["Profit_Margin"] < 0).sum(),
    "Zero": (df_clean["Profit_Margin"] == 0).sum(),
    "Positive": (df_clean["Profit_Margin"] > 0).sum()
}))

# Inspect sample rows with negative margins
print(df_clean.loc[
    df_clean["Profit_Margin"] < 0,
    ["Order_Value", "Discount_Applied", "Final_Amount", "Profit_Margin"]
].head(10))
```

3.3 Correcting invalid values (ratings > 5, negative profit margin)

```python
print("Restaurant ratings below 1:",
      (df_clean["Restaurant_Rating"] < 1).sum())

print("Restaurant ratings above 5:",
      (df_clean["Restaurant_Rating"] > 5).sum())
```

Correcting restaurant ratings

```python
# Cap restaurant ratings at the maximum allowed value of 5
df_clean["Restaurant_Rating"] = (
    df_clean["Restaurant_Rating"].clip(upper=5)
)

# Verify the correction
print("Ratings below 1:",
      (df_clean["Restaurant_Rating"] < 1).sum())

print("Ratings above 5:",
      (df_clean["Restaurant_Rating"] > 5).sum())
```
